```python
import pandas as pd
import requests

url = "https://fantasy.premierleague.com/api/bootstrap-static/"
data = requests.get(url).json()

print(data.keys())
```

```python
players = pd.DataFrame(data["elements"])
teams = pd.DataFrame(data["teams"])

print(players.shape)
players[["web_name", "team", "element_type", "now_cost", "total_points", "form", "minutes"]].head(10)
```

```python
players = players.merge(teams[["id", "name"]], left_on="team", right_on="id", suffixes=("", "_team"))
players["price"] = players["now_cost"] / 10
players["position"] = players["element_type"].map({1: "GK", 2: "DEF", 3: "MID", 4: "FWD"})

players[["web_name", "name", "position", "price", "total_points"]].sort_values("total_points", ascending=False).head(10)
```

```python
hist_url = "https://raw.githubusercontent.com/vaastav/Fantasy-Premier-League/master/data/2023-24/gws/merged_gw.csv"
hist = pd.read_csv(hist_url)

print(hist.shape)
hist.head()
```
